```python
import dai
import pandas as pd

bsd = '2024-12-30'
sd = '2025-01-01'
ed = '2026-05-01'

df = dai.query('SELECT date, instrument, close/m_lag(close, 1) as factor FROM cn_stock_bar1d', filters={'date': [bsd, ed]}).df()
df = df[(df['date']>=sd) & (df['date']<=ed)]

index_df = dai.query(
    "SELECT date, member_code as instrument FROM cn_stock_index_component WHERE instrument = '000852.SH'",
    filters={"date": [sd, ed]},
).df()

df = pd.merge(df, index_df, how='right', on=['date', 'instrument'])
df
```
