```python
import sys
path = "/home/aiuser/work/workspace/BigAlpha/data"
if path not in sys.path:
    sys.path.append(path)

from bigalpha_2026_financial.builder import Bigalpha2026FinancialBuilder
from utils import generate_dates_series
from datetime import datetime

dates_series = generate_dates_series(2019, 2024, "yearly")
dates_series.sort()
total_periods = len(dates_series)
for index, (start_date, end_date) in enumerate(dates_series, start=1):
    now = datetime.now()
    Bigalpha2026FinancialBuilder(start_date, end_date).build()
    duration = datetime.now() - now
    print(f"[{index}/{total_periods}] {100*index/total_periods:.2f}% | {start_date} 至 {end_date} | 耗时: {duration}")
```
