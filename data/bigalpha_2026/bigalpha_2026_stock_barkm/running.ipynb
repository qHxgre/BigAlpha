```python
import sys
path = "/home/aiuser/work/workspace/BigAlpha/data"
if path not in sys.path:
    sys.path.append(path)

from bigalpha_2026_stock_barkm.builder import Bigalpha2026StockBarKmBuilder
from utils import generate_dates_series
from datetime import datetime

K = 1

for K in [1, 5, 15, 30]:
    dates_series = generate_dates_series(2026, 2026, "monthly")
    dates_series.sort()
    total_periods = len(dates_series)
    for index, (start_date, end_date) in enumerate(dates_series, start=1):
        if start_date < '2026-08-01':
            continue
        if end_date > '2026-09-01':
            continue
        now = datetime.now()
        Bigalpha2026StockBarKmBuilder(start_date, end_date, K=K, suffix='private').build()
        duration = datetime.now() - now
        print(f"{K} [{index}/{total_periods}] {100*index/total_periods:.2f}% | {start_date} 至 {end_date} | 耗时: {duration}")
```
