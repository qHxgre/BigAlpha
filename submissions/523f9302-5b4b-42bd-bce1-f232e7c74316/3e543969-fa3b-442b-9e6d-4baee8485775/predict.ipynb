```python
"""BigAlpha final exact-PIT Stage-A inference notebook."""

import os
import pandas as pd

from train import MODEL_PATH, predict_submission_scores, train_and_save


def main(datasources, start_date, end_date):
    """Return exactly date, instrument, score for the requested interval."""
    if not os.path.exists(MODEL_PATH):
        raise FileNotFoundError(f"missing submission weights: {MODEL_PATH}")
    result = predict_submission_scores(
        datasources=datasources,
        start_date=str(start_date),
        end_date=str(end_date),
        model_path=MODEL_PATH,
    )
    result = result[["date", "instrument", "score"]].copy()
    result["date"] = pd.to_datetime(result["date"]).dt.normalize()
    return result
```
