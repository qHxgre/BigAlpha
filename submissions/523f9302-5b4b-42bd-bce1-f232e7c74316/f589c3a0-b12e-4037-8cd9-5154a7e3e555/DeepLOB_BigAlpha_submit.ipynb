# DeepLOB BigAlpha Submission
This notebook exposes the official `main(datasources, start_date, end_date)` inference entry and the `train_and_save(datasources)` private-board retraining entry.


```python
from bigalpha_deeplob import main as _main, train_and_save

def main(datasources, start_date, end_date):
    return _main(datasources, start_date, end_date)

assert callable(main)
assert callable(train_and_save)
```

```python
result = main(
    {"bar5m": "bigalpha_2026_stock_bar5m"},
    "2024-03-01",
    "2024-03-05",
)

print(result.head())
print("Shape:", result.shape)
print("Columns:", result.columns.tolist())
print("NaN:", result["score"].isna().sum())
print(
    "Duplicates:",
    result.duplicated(["date", "instrument"]).sum(),
)
print(result.groupby("date").size())
print(result.groupby("date")["score"].std())
```
