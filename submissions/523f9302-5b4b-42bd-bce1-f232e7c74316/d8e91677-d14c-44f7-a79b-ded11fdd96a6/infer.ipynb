```python
import importlib.util
import os


def _load_submission_runtime():
    candidates = []
    if "__file__" in globals():
        candidates.append(os.path.dirname(os.path.abspath(__file__)))
    candidates.append(os.getcwd())
    for directory in dict.fromkeys(candidates):
        train_path = os.path.join(directory, "train.py")
        model_path = os.path.join(directory, "model.json")
        if os.path.isfile(train_path) and os.path.isfile(model_path):
            spec = importlib.util.spec_from_file_location(
                "bigalpha_submission_train", train_path)
            module = importlib.util.module_from_spec(spec)
            spec.loader.exec_module(module)
            return module, model_path
    raise FileNotFoundError(
        "train.py and model.json must be next to infer.ipynb or in the working directory")


_runtime, _model_path = _load_submission_runtime()


def main(datasources, start_date, end_date):
    return _runtime.infer_scores(
        datasources, start_date, end_date, model_path=_model_path)
```
