# BigAlpha E2E 1-minute Transformer submission

The platform imports and calls `main(datasources, start_date, end_date)`.

```python
import os
import transformer_train_local as pipeline

MODEL_PATH = os.path.join(os.path.dirname(os.path.abspath(pipeline.__file__)), 'transformer_model.json')

def main(datasources, start_date, end_date):
    return pipeline.cloud_predict_scores(
        datasources, start_date, end_date, model_path=MODEL_PATH,
        instrument_chunk_size=40, batch_size=256,
    )
```
