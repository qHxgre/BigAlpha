```python
"""Self-contained submission factor: 条件波动与K线形态."""

SATELLITE_MODEL = ""
BASE_MODEL = ""
BASE_WEIGHT = 0.7
SATELLITE_WEIGHT = 0.3


def _decode_model(payload):
    import base64
    import json
    import zlib
    return json.loads(zlib.decompress(base64.b64decode(payload)).decode("utf-8"))


def _tree_predict(frame, model):
    import numpy as np

    names = model["feature_names"]
    x = frame.reindex(columns=names, fill_value=0.0).fillna(0.0).to_numpy(dtype="float64")
    result = np.zeros(len(frame), dtype="float64")
    for tree in model["tree_info"]:
        todo = [(tree["tree_structure"], np.arange(len(frame)))]
        while todo:
            node, rows = todo.pop()
            if not len(rows):
                continue
            if "leaf_value" in node:
                result[rows] += float(node["leaf_value"])
                continue
            values = x[rows, int(node["split_feature"])]
            threshold = float(node["threshold"])
            left_mask = values <= threshold
            # The exported models were fitted after fillna(0), but preserve the
            # LightGBM default branch for any future missing value.
            if node.get("default_left", True):
                left_mask = left_mask | ~np.isfinite(values)
            todo.append((node["left_child"], rows[left_mask]))
            todo.append((node["right_child"], rows[~left_mask]))
    return result


def _add_bar_features(daily):
    import numpy as np

    daily = daily.sort_values(["instrument", "date"]).copy()
    key = "instrument"
    daily["bar_ret1"] = daily["close"] / daily["pre_close"] - 1.0
    daily["bar_gap"] = daily["open"] / daily["pre_close"] - 1.0
    daily["bar_body"] = daily["close"] / daily["open"] - 1.0
    daily["bar_range"] = daily["high"] / daily["low"] - 1.0
    span = daily["high"] - daily["low"]
    daily["bar_close_location"] = (daily["close"] - daily["low"]) / (span + 1e-8)
    daily["bar_upper_wick"] = (daily["high"] - daily[["open", "close"]].max(axis=1)) / (span + 1e-8)
    daily["bar_lower_wick"] = (daily[["open", "close"]].min(axis=1) - daily["low"]) / (span + 1e-8)
    log_amount = np.log1p(daily["amount"].clip(lower=0))
    log_volume = np.log1p(daily["volume"].clip(lower=0))
    for window in (3, 5, 10, 20, 60):
        minimum = max(2, window // 2)
        daily[f"bar_ret_{window}"] = daily.groupby(key, group_keys=False)["close"].pct_change(window, fill_method=None)
        daily[f"bar_vol_{window}"] = daily.groupby(key, group_keys=False)["bar_ret1"].transform(
            lambda s: s.rolling(window, min_periods=minimum).std()
        )
        amount_mean = log_amount.groupby(daily[key]).transform(
            lambda s: s.rolling(window, min_periods=minimum).mean()
        )
        amount_std = log_amount.groupby(daily[key]).transform(
            lambda s: s.rolling(window, min_periods=minimum).std()
        )
        volume_mean = log_volume.groupby(daily[key]).transform(
            lambda s: s.rolling(window, min_periods=minimum).mean()
        )
        volume_std = log_volume.groupby(daily[key]).transform(
            lambda s: s.rolling(window, min_periods=minimum).std()
        )
        rolling_high = daily.groupby(key, group_keys=False)["high"].transform(
            lambda s: s.rolling(window, min_periods=minimum).max()
        )
        rolling_low = daily.groupby(key, group_keys=False)["low"].transform(
            lambda s: s.rolling(window, min_periods=minimum).min()
        )
        daily[f"bar_amount_z_{window}"] = (log_amount - amount_mean) / (amount_std + 1e-8)
        daily[f"bar_volume_z_{window}"] = (log_volume - volume_mean) / (volume_std + 1e-8)
        daily[f"bar_price_position_{window}"] = (daily["close"] - rolling_low) / (rolling_high - rolling_low + 1e-8)
    return daily


def main(datasources, start_date, end_date):
    import numpy as np
    import pandas as pd
    import dai

    bar1m = datasources["bar1m"]
    q_start = (pd.to_datetime(start_date) - pd.Timedelta(days=180)).strftime("%Y-%m-%d %H:%M:%S")
    sql_daily = f"""
        SELECT
            date::DATE::DATETIME AS date,
            instrument::STRING AS instrument,
            ARG_MIN(open, date) AS open,
            MAX(high) AS high,
            MIN(low) AS low,
            LAST(close ORDER BY date) AS close,
            LAST(volume ORDER BY date) AS volume,
            LAST(amount ORDER BY date) AS amount,
            ANY_VALUE(pre_close) AS pre_close
        FROM {bar1m}
        WHERE close > 0
        GROUP BY date::DATE, instrument
        ORDER BY instrument, date
    """
    daily = dai.query(sql_daily, filters={"date": [q_start, end_date]}, compression=True).df()
    daily["date"] = pd.to_datetime(daily["date"])
    daily = _add_bar_features(daily)
    factorlib = dai.query(
        "SELECT * FROM bigalpha_2026_factorlib",
        filters={"date": [q_start, end_date]}, compression=True,
    ).df()
    factorlib["date"] = pd.to_datetime(factorlib["date"])
    rename = {
        column: f"fl_{column}"
        for column in factorlib.columns
        if column not in {"date", "instrument", "close", "volume", "amount", "daily_return"}
    }
    factorlib = factorlib.rename(columns=rename)
    df = daily.merge(factorlib, on=["date", "instrument"], how="inner")
    satellite = _tree_predict(df, _decode_model(SATELLITE_MODEL))
    if BASE_MODEL:
        base = _tree_predict(df, _decode_model(BASE_MODEL))
        base_z = pd.Series(base, index=df.index).groupby(df["date"]).transform(
            lambda s: (s - s.mean()) / (s.std() + 1e-12)
        )
        satellite_z = pd.Series(satellite, index=df.index).groupby(df["date"]).transform(
            lambda s: (s - s.mean()) / (s.std() + 1e-12)
        )
        df["factor"] = BASE_WEIGHT * base_z + SATELLITE_WEIGHT * satellite_z
    else:
        df["factor"] = pd.Series(satellite, index=df.index).groupby(df["date"]).transform(
            lambda s: (s - s.mean()) / (s.std() + 1e-12)
        )
    s, e = pd.to_datetime(start_date), pd.to_datetime(end_date)
    df = df[(df["date"] >= s) & (df["date"] <= e)].copy()
    pool = dai.query(
        "SELECT date, instrument FROM bigalpha_2026_instruments",
        filters={"date": [start_date, end_date]},
    ).df()
    pool["date"] = pd.to_datetime(pool["date"])
    df["instrument"] = df["instrument"].astype(str)
    pool["instrument"] = pool["instrument"].astype(str)
    df = pd.merge(df, pool, on=["date", "instrument"], how="inner")
    df["factor"] = df["factor"].replace([np.inf, -np.inf], np.nan)
    return df[["date", "instrument", "factor"]]
```
