```python
def main(datasources, start_date, end_date):
    """组合因子 D v5：与 v3 正交的低相关因子等权组合
    成员: cfo_ta + cash_ratio + doji + overnight_gap + bigorder + depth_asym
    权重: 等权 1/6（全部与 combo_b_v3 低相关 |r|<0.25，组内无未来函数）
    """
    import pandas as pd
    import numpy as np
    import dai

    def _m_cfo(datasources, start_date, end_date):
        import dai

        fin = datasources["financial"]

        LOOKBACK_DAYS = 500
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH ttm_data AS (
            SELECT date, instrument, net_cffoa
            FROM {fin}
            WHERE category = 'ttm' AND shift = 0
        ),
        lf_data AS (
            SELECT date, instrument, total_assets
            FROM {fin}
            WHERE category = 'lf' AND shift = 0
        )
        SELECT
            CAST(t.date AS DATETIME) AS date,
            t.instrument,
            t.net_cffoa * 1.0 / NULLIF(l.total_assets, 0) AS factor
        FROM ttm_data t
        LEFT JOIN lf_data l ON t.date = l.date AND t.instrument = l.instrument
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()
        if df.empty:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        df['date'] = pd.to_datetime(df['date']).dt.tz_localize(None).astype('datetime64[ns]')
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        df = df.sort_values(['date'])

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [query_start, end_date]},
        ).df()
        if stk_pool.empty:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        stk_pool['date'] = pd.to_datetime(stk_pool['date']).dt.tz_localize(None).astype('datetime64[ns]')

        frames = []
        for inst in sorted(df['instrument'].unique()):
            fin_g = df[df['instrument'] == inst][['date', 'factor']].sort_values('date')
            pool_g = stk_pool[stk_pool['instrument'] == inst].sort_values('date')
            if pool_g.empty:
                continue
            merged = pd.merge_asof(
                pool_g,
                fin_g,
                on='date',
                direction='backward',
                allow_exact_matches=True,
            )
            frames.append(merged)

        if not frames:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        out = pd.concat(frames, ignore_index=True)

        start_ts = pd.to_datetime(start_date).tz_localize(None)
        end_ts = pd.to_datetime(end_date).tz_localize(None)
        out = out[(out['date'] >= start_ts) & (out['date'] <= end_ts)]
        out = out.dropna(subset=['factor'])
        out = out.sort_values(['date', 'instrument']).reset_index(drop=True)

        return out


    def _m_cash(datasources, start_date, end_date):
        import dai

        fin = datasources["financial"]

        LOOKBACK_DAYS = 500
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        SELECT
            CAST(date AS DATETIME) AS date,
            instrument,
            net_cffoa * 1.0 / NULLIF(net_profit, 0) AS factor
        FROM {fin}
        WHERE category = 'ttm' AND shift = 0 AND net_profit > 0
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()
        if df.empty:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        df['date'] = pd.to_datetime(df['date']).dt.tz_localize(None).astype('datetime64[ns]')
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        df = df.sort_values(['date'])

        # 交易日序列（股票池）
        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [query_start, end_date]},
        ).df()
        if stk_pool.empty:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        stk_pool['date'] = pd.to_datetime(stk_pool['date']).dt.tz_localize(None).astype('datetime64[ns]')

        # 用 merge_asof 按 instrument 分别匹配（最近公告日 backward）
        frames = []
        for inst in sorted(df['instrument'].unique()):
            fin_g = df[df['instrument'] == inst][['date', 'factor']].sort_values('date')
            pool_g = stk_pool[stk_pool['instrument'] == inst].sort_values('date')
            if pool_g.empty:
                continue
            merged = pd.merge_asof(
                pool_g,
                fin_g,
                on='date',
                direction='backward',
                allow_exact_matches=True,
            )
            frames.append(merged)

        if not frames:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        out = pd.concat(frames, ignore_index=True)

        # 裁剪到目标区间
        start_ts = pd.to_datetime(start_date).tz_localize(None)
        end_ts = pd.to_datetime(end_date).tz_localize(None)
        out = out[(out['date'] >= start_ts) & (out['date'] <= end_ts)]
        out = out.dropna(subset=['factor'])
        out = out.sort_values(['date', 'instrument']).reset_index(drop=True)

        # 负向因子取反，确保与未来收益正相关（IC>0）
        out['factor'] = -1.0 * out['factor']

        return out


    def _m_doji(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH daily AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                FIRST_VALUE(open) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date) AS open_px,
                LAST_VALUE(close) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING) AS close_px,
                MAX(high) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d')) AS high_px,
                MIN(low) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d')) AS low_px
            FROM {bar1m}
        ),
        doji AS (
            SELECT DISTINCT
                trading_day, instrument,
                CASE WHEN high_px > low_px
                     THEN ABS(close_px - open_px) * 1.0 / (high_px - low_px)
                     ELSE NULL END AS doji_ratio
            FROM daily
        ),
        doji_flag AS (
            SELECT
                trading_day, instrument,
                CASE WHEN doji_ratio IS NOT NULL AND doji_ratio < 0.1 THEN 1.0 ELSE 0.0 END AS is_doji
            FROM doji
        ),
        doji_freq AS (
            SELECT
                trading_day, instrument, is_doji,
                AVG(is_doji) OVER (PARTITION BY instrument ORDER BY trading_day ROWS BETWEEN 19 PRECEDING AND CURRENT ROW) AS factor
            FROM doji_flag
        )
        SELECT CAST(trading_day AS DATETIME) AS date, instrument, factor
        FROM doji_freq
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    def _m_gap(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH daily AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                FIRST_VALUE(open) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date) AS open_px,
                FIRST_VALUE(pre_close) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date) AS pre_close_px
            FROM {bar1m}
        ),
        daily_gap AS (
            SELECT DISTINCT
                trading_day, instrument,
                (open_px - pre_close_px) * 1.0 / NULLIF(pre_close_px, 0) AS gap
            FROM daily
            WHERE pre_close_px > 0
        ),
        gap_avg AS (
            SELECT
                trading_day, instrument, gap,
                AVG(gap) OVER (PARTITION BY instrument ORDER BY trading_day ROWS BETWEEN 9 PRECEDING AND CURRENT ROW) AS factor
            FROM daily_gap
        )
        SELECT CAST(trading_day AS DATETIME) AS date, instrument, factor
        FROM gap_avg
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    def _m_big(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        SELECT
            date, instrument, amount, deal_number, close
        FROM {bar1m}
        WHERE deal_number > 0 AND close > 0 AND amount > 0
        """
        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        # 分钟均笔成交额
        df['avg_size'] = df['amount'] / df['deal_number'].clip(lower=1)
        df['trading_day'] = pd.to_datetime(df['date']).dt.strftime('%Y-%m-%d')

        # 当日价格方向（分钟收盘价变化）
        df = df.sort_values(['instrument', 'date'])
        df['prev_close'] = df.groupby('instrument')['close'].shift(1)
        df['direction'] = np.sign(df['close'] - df['prev_close']).fillna(0)

        # 截面80分位阈值（按交易日全市场）
        thr = df.groupby('trading_day')['avg_size'].transform(
            lambda x: x.quantile(0.8))

        # 大单净流入（占比）
        df['big_flow'] = np.where(df['avg_size'] > thr, df['direction'] * df['amount'], 0)
        df['is_big'] = (df['avg_size'] > thr).astype(float)

        daily = df.groupby(['trading_day', 'instrument']).agg(
            net_flow=('big_flow', 'sum'),
            total_amt=('amount', 'sum'),
        ).reset_index()
        daily['factor'] = daily['net_flow'] / daily['total_amt'].clip(lower=1e-8)

        out = daily[['trading_day', 'instrument', 'factor']].rename(columns={'trading_day': 'date'})
        out['date'] = pd.to_datetime(out['date'])

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()
        stk_pool['date'] = pd.to_datetime(stk_pool['date'])

        out = pd.merge(out, stk_pool, how='inner', on=['date', 'instrument'])
        out = out.dropna(subset=['factor'])
        out = out[~out['factor'].isin([np.inf, -np.inf])]
        out = out.sort_values(['date', 'instrument']).reset_index(drop=True)

        # 负向因子取反，确保与未来收益正相关（IC>0）
        out['factor'] = -1.0 * out['factor']

        return out


    def _m_depth(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH cte AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                (bid_volume1 + bid_volume2 + bid_volume3 + bid_volume4 + bid_volume5 -
                 ask_volume1 - ask_volume2 - ask_volume3 - ask_volume4 - ask_volume5) * 1.0 / NULLIF(
                    bid_volume1 + bid_volume2 + bid_volume3 + bid_volume4 + bid_volume5 +
                    ask_volume1 + ask_volume2 + ask_volume3 + ask_volume4 + ask_volume5, 0
                ) AS da,
                amount
            FROM {bar1m}
        )
        SELECT
            CAST(trading_day AS DATETIME) AS date,
            instrument,
            SUM(da * amount) / NULLIF(SUM(amount), 0) AS factor
        FROM cte
        GROUP BY trading_day, instrument
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        # 负向因子取反，确保与未来收益正相关（IC>0）
        df['factor'] = -1.0 * df['factor']

        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    # ===== 计算 6 个成员因子 =====
    f1 = _m_cfo(datasources, start_date, end_date).rename(columns={'factor': 'f_cfo'})
    f2 = _m_cash(datasources, start_date, end_date).rename(columns={'factor': 'f_cash'})
    f3 = _m_doji(datasources, start_date, end_date).rename(columns={'factor': 'f_doji'})
    f4 = _m_gap(datasources, start_date, end_date).rename(columns={'factor': 'f_gap'})
    f5 = _m_big(datasources, start_date, end_date).rename(columns={'factor': 'f_big'})
    f6 = _m_depth(datasources, start_date, end_date).rename(columns={'factor': 'f_depth'})

    # ===== 组合（union 对齐 + ffill + z-score + 等权）=====
    for df in [f1, f2, f3, f4, f5, f6]:
        df['date'] = pd.to_datetime(df['date']).astype('datetime64[ns]')

    frames = {}
    for name, df in [('f_cfo', f1), ('f_cash', f2), ('f_doji', f3), ('f_gap', f4), ('f_big', f5), ('f_depth', f6)]:
        frames[name] = df.set_index(['date', 'instrument'])[name]
    panel = pd.DataFrame(frames)

    def ff(series):
        return series.groupby(level='instrument').ffill()
    filled = panel.apply(ff)

    def zscore(series):
        return series.groupby(level='date').transform(
            lambda x: (x - x.mean()) / (x.std() + 1e-10))
    zs = filled.apply(zscore)

    weights = {'f_cfo': 1/6, 'f_cash': 1/6, 'f_doji': 1/6, 'f_gap': 1/6, 'f_big': 1/6, 'f_depth': 1/6}
    combo = sum(zs[col] * w for col, w in weights.items())

    out = combo.reset_index(name='factor')
    out = out.dropna(subset=['factor'])
    out = out[~out['factor'].isin([np.inf, -np.inf])]

    stk_pool = dai.query(
        "SELECT date, instrument FROM bigalpha_2026_instruments",
        filters={'date': [start_date, end_date]},
    ).df()
    stk_pool['date'] = pd.to_datetime(stk_pool['date']).astype('datetime64[ns]')
    out = pd.merge(out, stk_pool, how='inner', on=['date', 'instrument'])
    out = out.dropna(subset=['factor'])
    out = out[~out['factor'].isin([np.inf, -np.inf])]
    out = out.sort_values(['date', 'instrument']).reset_index(drop=True)

    return out

if __name__ == '__main__':
    from bigmodule import M
    import dai
    import structlog

    logger = structlog.get_logger()

    datasources = {'bar1m': 'bigalpha_2026_stock_bar1m', 'financial': 'bigalpha_2026_financial'}
    start_date = '2024-01-01 00:00:00'
    end_date = '2024-12-31 23:59:59'

    logger.info(f"计算因子，区间：{start_date} ~ {end_date}")
    factor_data = main(datasources, start_date, end_date)

    logger.info(f"读取因子库，区间：{start_date} ~ {end_date}")
    factor_pool = dai.query(
        "SELECT * FROM bigalpha_2026_factorlib",
        filters={'date': [start_date, end_date]},
    ).df()

    result = M.bigalpha_eval._latest(
        factor_data=factor_data,
        factor_pool=factor_pool,
        process_pools=False,
        show=True,
    )
```
