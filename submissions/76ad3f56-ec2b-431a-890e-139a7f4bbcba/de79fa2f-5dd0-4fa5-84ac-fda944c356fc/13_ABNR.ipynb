```python
def main(datasources, start_date, end_date):

    import numpy as np
    import pandas as pd
    import dai


    # =====================================================
    # 0. 参数设置
    # =====================================================

    bar_table = datasources.get(
        "bar1d",
        "cn_stock_bar1d"
    )

    status_table = datasources.get(
        "status",
        "cn_stock_status"
    )

    basic_info_table = datasources.get(
        "basic_info",
        "cn_stock_basic_info"
    )

    pool_table = datasources.get(
        "pool",
        "bigalpha_2026_instruments"
    )


    # 当前负向反转频率窗口
    short_window = int(
        datasources.get(
            "short_window",
            20
        )
    )

    # 历史正常频率窗口
    history_window = int(
        datasources.get(
            "history_window",
            252
        )
    )

    # 当前窗口最少有效交易日
    min_short_obs = int(
        datasources.get(
            "min_short_obs",
            15
        )
    )

    # 历史窗口最少有效交易日
    min_history_obs = int(
        datasources.get(
            "min_history_obs",
            200
        )
    )

    # 因子方向
    # 1.0表示使用原始AB_NR
    # -1.0表示使用相反方向
    factor_sign = float(
        datasources.get(
            "factor_sign",
            1.0
        )
    )

    # 截面缩尾分位数
    winsor_lower = float(
        datasources.get(
            "winsor_lower",
            0.01
        )
    )

    winsor_upper = float(
        datasources.get(
            "winsor_upper",
            0.99
        )
    )

    # 截面标准化最少样本数
    min_cs_obs = int(
        datasources.get(
            "min_cs_obs",
            30
        )
    )

    eps = 1e-12


    if short_window < 2:
        raise ValueError(
            "short_window必须大于等于2"
        )

    if history_window < short_window:
        raise ValueError(
            "history_window应大于short_window"
        )

    if min_short_obs > short_window:
        raise ValueError(
            "min_short_obs不能大于short_window"
        )

    if min_history_obs > history_window:
        raise ValueError(
            "min_history_obs不能大于history_window"
        )


    # =====================================================
    # 1. 日期处理
    # =====================================================

    start_ts = pd.to_datetime(
        start_date
    )

    end_ts = pd.to_datetime(
        end_date
    )

    if start_ts > end_ts:
        raise ValueError(
            "start_date不能晚于end_date"
        )


    start_day = start_ts.normalize()
    end_day = end_ts.normalize()


    # 为20日当前窗口和252日历史窗口预留数据
    total_required_days = (
        short_window
        +
        history_window
    )

    calendar_buffer_days = int(
        total_required_days * 1.8
        +
        90
    )

    query_start_day = (
        start_day
        -
        pd.Timedelta(
            days=calendar_buffer_days
        )
    )


    print(
        f"[ABNR-Daily] start:       "
        f"{start_day}"
    )

    print(
        f"[ABNR-Daily] end:         "
        f"{end_day}"
    )

    print(
        f"[ABNR-Daily] query start: "
        f"{query_start_day}"
    )

    print(
        f"[ABNR-Daily] short window: "
        f"{short_window}"
    )

    print(
        f"[ABNR-Daily] history window: "
        f"{history_window}"
    )


    # =====================================================
    # 2. 通用函数
    # =====================================================

    def clean_instrument(series):

        return (
            series
            .astype(str)
            .str.strip()
            .str.upper()
        )


    def cs_zscore(series):

        """
        每日截面处理：

        1. 无穷值转换为缺失值；
        2. 1%和99%分位数缩尾；
        3. 截面z-score标准化。
        """

        x = pd.to_numeric(
            series,
            errors="coerce"
        )

        x = x.replace(
            [
                np.inf,
                -np.inf
            ],
            np.nan
        )

        valid = x.dropna()

        if len(valid) < min_cs_obs:

            return pd.Series(
                np.nan,
                index=x.index,
                dtype=float
            )


        lower = valid.quantile(
            winsor_lower
        )

        upper = valid.quantile(
            winsor_upper
        )


        clipped = x.clip(
            lower=lower,
            upper=upper
        )


        mean_value = clipped.mean()

        std_value = clipped.std(
            ddof=0
        )


        if (
            pd.isna(std_value)
            or
            std_value < eps
        ):

            result = pd.Series(
                np.nan,
                index=x.index,
                dtype=float
            )

            result.loc[
                clipped.notna()
            ] = 0.0

            return result


        return (
            clipped
            -
            mean_value
        ) / std_value


    # =====================================================
    # 3. 查询日频行情数据
    # =====================================================

    print(
        "[ABNR-Daily] 查询日频行情"
    )


    query = f"""
    SELECT
        b.date,
        b.instrument,
        b.open,
        b.close,
        b.pre_close,
        b.adjust_factor,
        s.suspended,
        s.price_limit_status,
        s.st_status,
        info.list_date
    FROM {bar_table} b
    JOIN {status_table} s
        USING (date, instrument)
    JOIN {basic_info_table} info
        USING (instrument)
    WHERE
        b.date >= info.list_date
        AND s.suspended = 0
    """


    data = dai.query(
        query,
        filters={
            "date": [
                query_start_day.strftime(
                    "%Y-%m-%d"
                ),
                end_day.strftime(
                    "%Y-%m-%d"
                )
            ]
        },
        compression=True
    ).df()


    print(
        f"[ABNR-Daily] 原始记录: "
        f"{len(data):,}"
    )


    if data.empty:
        raise RuntimeError(
            "日频行情数据为空"
        )


    # =====================================================
    # 4. 基础清洗
    # =====================================================

    print(
        "[ABNR-Daily] 数据清洗"
    )


    data["date"] = (
        pd.to_datetime(
            data["date"],
            errors="coerce"
        )
        .dt
        .normalize()
    )


    data["instrument"] = clean_instrument(
        data["instrument"]
    )


    data["list_date"] = pd.to_datetime(
        data["list_date"],
        errors="coerce"
    )


    numeric_cols = [
        "open",
        "close",
        "pre_close",
        "adjust_factor",
        "suspended",
        "price_limit_status",
        "st_status"
    ]


    for col in numeric_cols:

        if col in data.columns:

            data[col] = pd.to_numeric(
                data[col],
                errors="coerce"
            )


    data = data.dropna(
        subset=[
            "date",
            "instrument",
            "open",
            "close"
        ]
    )


    data = data.loc[
        (
            data["open"] > 0
        )
        &
        (
            data["close"] > 0
        )
    ].copy()


    # 删除完全重复记录
    exact_before = len(data)

    data = data.drop_duplicates()


    print(
        f"[ABNR-Daily] 删除完全重复记录: "
        f"{exact_before - len(data):,}"
    )


    # 同一股票同一天只保留最后一条
    duplicate_count = data.duplicated(
        subset=[
            "date",
            "instrument"
        ],
        keep=False
    ).sum()


    print(
        f"[ABNR-Daily] 日期股票重复记录: "
        f"{duplicate_count:,}"
    )


    data = (
        data
        .sort_values(
            [
                "instrument",
                "date"
            ]
        )
        .drop_duplicates(
            subset=[
                "date",
                "instrument"
            ],
            keep="last"
        )
        .reset_index(
            drop=True
        )
    )


    # =====================================================
    # 5. 构造复权价格
    # =====================================================

    print(
        "[ABNR-Daily] 构造复权价格"
    )


    # adjust_factor异常时使用1
    data["adjust_factor"] = (
        data["adjust_factor"]
        .replace(
            [
                np.inf,
                -np.inf
            ],
            np.nan
        )
    )


    data.loc[
        data["adjust_factor"] <= 0,
        "adjust_factor"
    ] = np.nan


    data["adjust_factor"] = (
        data.groupby(
            "instrument",
            observed=True
        )["adjust_factor"]
        .ffill()
        .fillna(1.0)
    )


    data["adj_open"] = (
        data["open"]
        *
        data["adjust_factor"]
    )


    data["adj_close"] = (
        data["close"]
        *
        data["adjust_factor"]
    )


    # 上一个有效交易日的复权收盘价
    data["adj_pre_close"] = (
        data.groupby(
            "instrument",
            observed=True
        )["adj_close"]
        .shift(1)
    )


    # =====================================================
    # 6. 构造隔夜和日内收益
    # =====================================================

    print(
        "[ABNR-Daily] 构造收益"
    )


    # 日内收益：开盘到收盘
    data["ret_oc"] = (
        data["adj_close"]
        /
        data["adj_open"]
        -
        1.0
    )


    # 隔夜收益：上一交易日收盘到当日开盘
    data["ret_co"] = (
        data["adj_open"]
        /
        data["adj_pre_close"]
        -
        1.0
    )


    # 收盘到收盘收益，仅用于诊断
    data["ret_cc"] = (
        data["adj_close"]
        /
        data["adj_pre_close"]
        -
        1.0
    )


    return_cols = [
        "ret_oc",
        "ret_co",
        "ret_cc"
    ]


    for col in return_cols:

        data[col] = (
            data[col]
            .replace(
                [
                    np.inf,
                    -np.inf
                ],
                np.nan
            )
        )


    # 排除明显异常收益
    # 该条件主要防止复权字段异常
    valid_return = (
        data["ret_oc"].notna()
        &
        data["ret_co"].notna()
        &
        data["ret_cc"].notna()
        &
        data["ret_oc"].between(
            -0.50,
            0.50
        )
        &
        data["ret_co"].between(
            -0.50,
            0.50
        )
    )


    print(
        f"[ABNR-Daily] 有效收益记录率: "
        f"{valid_return.mean():.2%}"
    )


    # =====================================================
    # 7. 定义负向反转事件
    # =====================================================

    print(
        "[ABNR-Daily] 识别负向反转事件"
    )


    # 保持原代码定义：
    #
    # 隔夜收益为正
    # 日内收益为负
    #
    # 表示隔夜上涨在日内发生负向修正

    negative_reversal = (
        (
            data["ret_co"] > 0
        )
        &
        (
            data["ret_oc"] < 0
        )
    )


    # 有效交易日：
    # 事件发生记为1
    # 事件未发生记为0
    # 收益无效记为NaN
    data["neg_rev"] = np.where(
        valid_return,
        negative_reversal.astype(float),
        np.nan
    )


    # 正向反转事件，仅保留作诊断
    positive_reversal = (
        (
            data["ret_co"] < 0
        )
        &
        (
            data["ret_oc"] > 0
        )
    )


    data["pos_rev"] = np.where(
        valid_return,
        positive_reversal.astype(float),
        np.nan
    )


    print(
        f"[ABNR-Daily] 负向反转事件率: "
        f"{data['neg_rev'].mean():.4f}"
    )


    # =====================================================
    # 8. 计算短期负向反转频率
    # =====================================================

    print(
        "[ABNR-Daily] 计算短期NR"
    )


    data["NR_short"] = (
        data.groupby(
            "instrument",
            observed=True
        )["neg_rev"]
        .transform(
            lambda x:
            x.rolling(
                window=short_window,
                min_periods=min_short_obs
            ).mean()
        )
    )


    data["short_valid_days"] = (
        data.groupby(
            "instrument",
            observed=True
        )["neg_rev"]
        .transform(
            lambda x:
            x.notna()
            .astype(int)
            .rolling(
                window=short_window,
                min_periods=1
            ).sum()
        )
    )


    # =====================================================
    # 9. 计算历史正常负向反转频率
    # =====================================================

    print(
        "[ABNR-Daily] 计算历史NR"
    )


    # shift(short_window)使历史窗口位于当前短期窗口之前
    #
    # 当前窗口：
    # t-short_window+1 至 t
    #
    # 历史窗口：
    # t-short_window-history_window+1
    # 至
    # t-short_window

    data["NR_hist_mean"] = (
        data.groupby(
            "instrument",
            observed=True
        )["neg_rev"]
        .transform(
            lambda x:
            x.shift(
                short_window
            )
            .rolling(
                window=history_window,
                min_periods=min_history_obs
            )
            .mean()
        )
    )


    data["history_valid_days"] = (
        data.groupby(
            "instrument",
            observed=True
        )["neg_rev"]
        .transform(
            lambda x:
            x.notna()
            .astype(int)
            .shift(
                short_window
            )
            .rolling(
                window=history_window,
                min_periods=1
            )
            .sum()
        )
    )


    # =====================================================
    # 10. 构造日度AB_NR
    # =====================================================

    print(
        "[ABNR-Daily] 构造日度AB_NR"
    )


    valid_abnr = (
        data["NR_short"].notna()
        &
        data["NR_hist_mean"].notna()
        &
        (
            data["NR_hist_mean"] > eps
        )
        &
        (
            data["short_valid_days"]
            >=
            min_short_obs
        )
        &
        (
            data["history_valid_days"]
            >=
            min_history_obs
        )
    )


    data["AB_NR_raw"] = np.where(
        valid_abnr,
        data["NR_short"]
        /
        data["NR_hist_mean"],
        np.nan
    )


    data["AB_NR_raw"] = (
        data["AB_NR_raw"]
        .replace(
            [
                np.inf,
                -np.inf
            ],
            np.nan
        )
    )


    print(
        f"[ABNR-Daily] AB_NR有效率: "
        f"{data['AB_NR_raw'].notna().mean():.2%}"
    )


    print(
        f"[ABNR-Daily] 历史频率为0比例: "
        f"{(data['NR_hist_mean'] <= eps).mean():.2%}"
    )


    # =====================================================
    # 11. 每日截面标准化
    # =====================================================

    print(
        "[ABNR-Daily] 每日截面标准化"
    )


    data["factor"] = (
        data.groupby(
            "date",
            observed=True
        )["AB_NR_raw"]
        .transform(
            cs_zscore
        )
    )


    data["factor"] = (
        factor_sign
        *
        data["factor"]
    )


    # =====================================================
    # 12. 截取正式样本区间
    # =====================================================

    factor_data = data.loc[
        (
            data["date"]
            >=
            start_day
        )
        &
        (
            data["date"]
            <=
            end_day
        ),
        [
            "date",
            "instrument",
            "factor"
        ]
    ].copy()


    factor_data = (
        factor_data
        .drop_duplicates(
            subset=[
                "date",
                "instrument"
            ],
            keep="last"
        )
        .sort_values(
            [
                "date",
                "instrument"
            ]
        )
        .reset_index(
            drop=True
        )
    )


    print(
        f"[ABNR-Daily] 正式因子记录: "
        f"{len(factor_data):,}"
    )


    # =====================================================
    # 13. 查询股票池
    # =====================================================

    print(
        "[ABNR-Daily] 查询股票池"
    )


    pool_query = f"""
    SELECT
        date,
        instrument
    FROM {pool_table}
    """


    pool = dai.query(
        pool_query,
        filters={
            "date": [
                start_day.strftime(
                    "%Y-%m-%d"
                ),
                end_day.strftime(
                    "%Y-%m-%d"
                )
            ]
        },
        compression=True
    ).df()


    if pool.empty:
        raise RuntimeError(
            "股票池为空"
        )


    pool["date"] = (
        pd.to_datetime(
            pool["date"],
            errors="coerce"
        )
        .dt
        .normalize()
    )


    pool["instrument"] = clean_instrument(
        pool["instrument"]
    )


    pool = (
        pool
        .dropna(
            subset=[
                "date",
                "instrument"
            ]
        )
        .drop_duplicates(
            subset=[
                "date",
                "instrument"
            ]
        )
    )


    print(
        f"[ABNR-Daily] 股票池记录: "
        f"{len(pool):,}"
    )


    # =====================================================
    # 14. 连接股票池
    # =====================================================

    result = pool.merge(
        factor_data,
        how="left",
        on=[
            "date",
            "instrument"
        ],
        validate="one_to_one"
    )


    coverage = (
        result["factor"]
        .notna()
        .mean()
    )


    print(
        f"[ABNR-Daily] 原始覆盖率: "
        f"{coverage:.2%}"
    )


    # =====================================================
    # 15. 缺失值处理
    # =====================================================

    result["factor"] = (
        result["factor"]
        .replace(
            [
                np.inf,
                -np.inf
            ],
            np.nan
        )
    )


    # 因子已经完成每日截面标准化
    # 缺失值填0表示当日截面中性水平
    result["factor"] = (
        result["factor"]
        .fillna(
            0.0
        )
        .astype(float)
    )


    # =====================================================
    # 16. 最终输出
    # =====================================================

    result = (
        result[
            [
                "date",
                "instrument",
                "factor"
            ]
        ]
        .sort_values(
            [
                "date",
                "instrument"
            ]
        )
        .reset_index(
            drop=True
        )
    )


    print(
        f"[ABNR-Daily] 输出完成: "
        f"{len(result):,}"
    )


    print(
        "[ABNR-Daily] 因子描述统计"
    )


    print(
        result["factor"]
        .describe()
    )


    print(
        "[ABNR-Daily] 前5行"
    )


    print(
        result.head()
    )


    return result





# =========================================================
# 本地运行与评估
# =========================================================

if __name__ == "__main__":

    from bigmodule import M
    import dai


    try:
        from IPython.display import display
    except ImportError:
        display = print


    # =====================================================
    # 数据源与参数
    # =====================================================

    datasources = {

        "bar1d":
        "cn_stock_bar1d",

        "status":
        "cn_stock_status",

        "basic_info":
        "cn_stock_basic_info",

        "pool":
        "bigalpha_2026_instruments",

        # 当前反转频率窗口
        "short_window":
        20,

        # 历史正常频率窗口
        "history_window":
        252,

        # 当前20日中至少15个有效交易日
        "min_short_obs":
        15,

        # 历史252日中至少200个有效交易日
        "min_history_obs":
        200,

        # 原始方向
        "factor_sign":
        1.0,

        # 每日截面缩尾
        "winsor_lower":
        0.01,

        "winsor_upper":
        0.99,

        # 每日截面最少股票数
        "min_cs_obs":
        30

    }


    start_date = (
        "2019-01-01 00:00:00"
    )


    end_date = (
        "2024-12-31 23:59:59"
    )


    print(
        "[ABNR-Daily] 开始计算"
    )


    factor_data = main(
        datasources=datasources,
        start_date=start_date,
        end_date=end_date
    )


    if factor_data.empty:
        raise RuntimeError(
            "ABNR-Daily返回为空"
        )


    display(
        factor_data["factor"]
        .describe()
    )


    display(
        factor_data.head(
            10
        )
    )


    print(
        "[ABNR-Daily] 开始评估"
    )


    factor_pool = dai.query(
        """
        SELECT *
        FROM bigalpha_2026_factorlib
        """,
        filters={
            "date": [
                start_date,
                end_date
            ]
        },
        compression=True
    ).df()


    evaluation_result = M.bigalpha_eval._latest(
        factor_data=factor_data,
        factor_pool=factor_pool,
        start_date=start_date,
        end_date=end_date,
        process_pools=False,
        show=True
    )


    display(
        evaluation_result
    )
```
