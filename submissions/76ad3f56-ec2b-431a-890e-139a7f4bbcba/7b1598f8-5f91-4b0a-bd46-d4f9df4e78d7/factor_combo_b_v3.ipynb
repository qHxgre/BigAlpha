```python
def main(datasources, start_date, end_date):
    """组合因子 B V3：ICIR 加权合成（真实因子实现内联）
    成员: obi1 + accruals + er + intraday_ret + rskew + pv_corr
    权重: ICIR 归一化 {obi1:0.089 acc:0.22 er:0.154 ir:0.157 skew:0.165 pvc:0.215}
    """
    import pandas as pd
    import numpy as np
    import dai

    def _m_obi1(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH cte AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                (bid_volume1 - ask_volume1) * 1.0 / NULLIF(bid_volume1 + ask_volume1, 0) AS obi,
                amount
            FROM {bar1m}
            WHERE bid_volume1 + ask_volume1 > 0
        )
        SELECT
            CAST(trading_day AS DATETIME) AS date,
            instrument,
            SUM(obi * amount) / NULLIF(SUM(amount), 0) AS factor
        FROM cte
        GROUP BY trading_day, instrument
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        # 负向因子取反，确保与未来收益正相关（IC>0）
        df['factor'] = -1.0 * df['factor']

        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    def _m_acc(datasources, start_date, end_date):
        import dai

        fin = datasources["financial"]

        LOOKBACK_DAYS = 500
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH ttm_data AS (
            SELECT date, instrument, net_profit, net_cffoa
            FROM {fin}
            WHERE category = 'ttm' AND shift = 0
        ),
        lf_data AS (
            SELECT date, instrument, total_assets
            FROM {fin}
            WHERE category = 'lf' AND shift = 0
        )
        SELECT
            CAST(t.date AS DATETIME) AS date,
            t.instrument,
            (t.net_profit - t.net_cffoa) * 1.0 / NULLIF(l.total_assets, 0) AS factor
        FROM ttm_data t
        LEFT JOIN lf_data l ON t.date = l.date AND t.instrument = l.instrument
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()
        if df.empty:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        df['date'] = pd.to_datetime(df['date']).dt.tz_localize(None).astype('datetime64[ns]')
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        df = df.sort_values(['date'])

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [query_start, end_date]},
        ).df()
        if stk_pool.empty:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        stk_pool['date'] = pd.to_datetime(stk_pool['date']).dt.tz_localize(None).astype('datetime64[ns]')

        frames = []
        for inst in sorted(df['instrument'].unique()):
            fin_g = df[df['instrument'] == inst][['date', 'factor']].sort_values('date')
            pool_g = stk_pool[stk_pool['instrument'] == inst].sort_values('date')
            if pool_g.empty:
                continue
            merged = pd.merge_asof(
                pool_g,
                fin_g,
                on='date',
                direction='backward',
                allow_exact_matches=True,
            )
            frames.append(merged)

        if not frames:
            return pd.DataFrame(columns=['date', 'instrument', 'factor'])
        out = pd.concat(frames, ignore_index=True)

        start_ts = pd.to_datetime(start_date).tz_localize(None)
        end_ts = pd.to_datetime(end_date).tz_localize(None)
        out = out[(out['date'] >= start_ts) & (out['date'] <= end_ts)]
        out = out.dropna(subset=['factor'])
        out = out.sort_values(['date', 'instrument']).reset_index(drop=True)

        # 负向因子取反，确保与未来收益正相关（IC>0）
        out['factor'] = -1.0 * out['factor']

        return out


    def _m_er(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH daily AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                close,
                LAG(close) OVER (PARTITION BY instrument ORDER BY date) AS prev_close
            FROM {bar1m}
        ),
        paths AS (
            SELECT
                trading_day, instrument,
                ABS(close - prev_close) AS path_len
            FROM daily
            WHERE prev_close IS NOT NULL
        ),
        path_sum AS (
            SELECT
                trading_day, instrument,
                SUM(path_len) AS total_path
            FROM paths
            GROUP BY trading_day, instrument
        ),
        daily_oc AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                FIRST_VALUE(open) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date) AS open_px,
                LAST_VALUE(close) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING) AS close_px
            FROM {bar1m}
        ),
        er_daily AS (
            SELECT DISTINCT
                d.trading_day, d.instrument,
                ABS(d.close_px - d.open_px) * 1.0 / NULLIF(p.total_path, 0) AS er
            FROM daily_oc d
            JOIN path_sum p ON d.trading_day = p.trading_day AND d.instrument = p.instrument
            WHERE p.total_path > 0
        ),
        er_avg AS (
            SELECT
                trading_day, instrument, er,
                AVG(er) OVER (PARTITION BY instrument ORDER BY trading_day ROWS BETWEEN 19 PRECEDING AND CURRENT ROW) AS factor
            FROM er_daily
        )
        SELECT CAST(trading_day AS DATETIME) AS date, instrument, factor
        FROM er_avg
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        # 负向因子取反，确保与未来收益正相关（IC>0）
        df['factor'] = -1.0 * df['factor']

        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    def _m_ir(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH daily AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                FIRST_VALUE(open) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date) AS open_px,
                LAST_VALUE(close) OVER (PARTITION BY instrument, strftime(date, '%Y-%m-%d') ORDER BY date ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING) AS close_px
            FROM {bar1m}
        ),
        daily_ret AS (
            SELECT DISTINCT
                trading_day, instrument,
                (close_px - open_px) * 1.0 / NULLIF(open_px, 0) AS ireturn
            FROM daily
            WHERE open_px > 0
        ),
        ret_avg AS (
            SELECT
                trading_day, instrument, ireturn,
                AVG(ireturn) OVER (PARTITION BY instrument ORDER BY trading_day ROWS BETWEEN 9 PRECEDING AND CURRENT ROW) AS factor
            FROM daily_ret
        )
        SELECT CAST(trading_day AS DATETIME) AS date, instrument, factor
        FROM ret_avg
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        # 负向因子取反，确保与未来收益正相关（IC>0）
        df['factor'] = -1.0 * df['factor']

        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    def _m_skew(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH returns AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                close / NULLIF(LAG(close) OVER (PARTITION BY instrument ORDER BY date), 0) - 1 AS ret
            FROM {bar1m}
            WHERE close > 0
        ),
        moments AS (
            SELECT
                trading_day, instrument,
                AVG(ret) AS m1,
                AVG(ret * ret) AS m2,
                AVG(ret * ret * ret) AS m3,
                COUNT(*) AS n
            FROM returns
            WHERE ret IS NOT NULL
            GROUP BY trading_day, instrument
        )
        SELECT
            CAST(trading_day AS DATETIME) AS date,
            instrument,
            CASE WHEN m2 > 0 THEN m3 / POWER(m2, 1.5) ELSE NULL END AS factor
        FROM moments
        WHERE n >= 60
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        # 负向因子取反，确保与未来收益正相关（IC>0）
        df['factor'] = -1.0 * df['factor']

        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    def _m_pvc(datasources, start_date, end_date):
        import dai

        bar1m = datasources["bar1m"]

        LOOKBACK_DAYS = 45
        query_start_date = pd.to_datetime(start_date) - pd.Timedelta(days=LOOKBACK_DAYS)
        query_start = query_start_date.strftime('%Y-%m-%d %H:%M:%S')

        sql = f"""
        WITH cte_daily AS (
            SELECT
                instrument,
                strftime(date, '%Y-%m-%d') AS trading_day,
                corr(close, volume) AS pv_corr
            FROM {bar1m}
            GROUP BY instrument, trading_day
        ),
        cte_avg AS (
            SELECT
                trading_day,
                instrument,
                pv_corr,
                AVG(pv_corr) OVER (
                    PARTITION BY instrument
                    ORDER BY trading_day
                    ROWS BETWEEN 19 PRECEDING AND CURRENT ROW
                ) AS pv_corr_avg
            FROM cte_daily
        )
        SELECT
            CAST(trading_day AS DATETIME) AS date,
            instrument,
            -1.0 * pv_corr_avg AS factor
        FROM cte_avg
        """

        df = dai.query(
            sql,
            filters={'date': [query_start, end_date]},
            compression=True,
        ).df()

        stk_pool = dai.query(
            "SELECT date, instrument FROM bigalpha_2026_instruments",
            filters={'date': [start_date, end_date]},
        ).df()

        df = pd.merge(df, stk_pool, how='inner', on=['date', 'instrument'])
        df = df.dropna(subset=['factor'])
        df = df[~df['factor'].isin([np.inf, -np.inf])]
        df = df.sort_values(['date', 'instrument']).reset_index(drop=True)

        return df


    # ===== 计算 6 个成员因子 =====
    f1 = _m_obi1(datasources, start_date, end_date).rename(columns={'factor': 'f_obi1'})
    f2 = _m_acc(datasources, start_date, end_date).rename(columns={'factor': 'f_acc'})
    f3 = _m_er(datasources, start_date, end_date).rename(columns={'factor': 'f_er'})
    f4 = _m_ir(datasources, start_date, end_date).rename(columns={'factor': 'f_ir'})
    f5 = _m_skew(datasources, start_date, end_date).rename(columns={'factor': 'f_skew'})
    f6 = _m_pvc(datasources, start_date, end_date).rename(columns={'factor': 'f_pvc'})

    # ===== 组合（与 build_combo_v3 一致：union 对齐 + ffill + z-score + ICIR 加权）=====
    for df in [f1, f2, f3, f4, f5, f6]:
        df['date'] = pd.to_datetime(df['date']).astype('datetime64[ns]')

    frames = {}
    for name, df in [('f_obi1', f1), ('f_acc', f2), ('f_er', f3), ('f_ir', f4), ('f_skew', f5), ('f_pvc', f6)]:
        frames[name] = df.set_index(['date', 'instrument'])[name]
    panel = pd.DataFrame(frames)

    def ff(series):
        return series.groupby(level='instrument').ffill()
    filled = panel.apply(ff)

    def zscore(series):
        return series.groupby(level='date').transform(
            lambda x: (x - x.mean()) / (x.std() + 1e-10))
    zs = filled.apply(zscore)

    weights = {'f_obi1': 0.089, 'f_acc': 0.22, 'f_er': 0.154, 'f_ir': 0.157, 'f_skew': 0.165, 'f_pvc': 0.215}
    combo = sum(zs[col] * w for col, w in weights.items())

    out = combo.reset_index(name='factor')
    out = out.dropna(subset=['factor'])
    out = out[~out['factor'].isin([np.inf, -np.inf])]

    stk_pool = dai.query(
        "SELECT date, instrument FROM bigalpha_2026_instruments",
        filters={'date': [start_date, end_date]},
    ).df()
    stk_pool['date'] = pd.to_datetime(stk_pool['date']).astype('datetime64[ns]')
    out = pd.merge(out, stk_pool, how='inner', on=['date', 'instrument'])
    out = out.dropna(subset=['factor'])
    out = out[~out['factor'].isin([np.inf, -np.inf])]
    out = out.sort_values(['date', 'instrument']).reset_index(drop=True)

    return out

if __name__ == '__main__':
    from bigmodule import M
    import dai
    import structlog

    logger = structlog.get_logger()

    datasources = {'bar1m': 'bigalpha_2026_stock_bar1m', 'financial': 'bigalpha_2026_financial'}
    start_date = '2024-01-01 00:00:00'
    end_date = '2024-12-31 23:59:59'

    logger.info(f"计算因子，区间：{start_date} ~ {end_date}")
    factor_data = main(datasources, start_date, end_date)

    logger.info(f"读取因子库，区间：{start_date} ~ {end_date}")
    factor_pool = dai.query(
        "SELECT * FROM bigalpha_2026_factorlib",
        filters={'date': [start_date, end_date]},
    ).df()

    result = M.bigalpha_eval._latest(
        factor_data=factor_data,
        factor_pool=factor_pool,
        process_pools=False,
        show=True,
    )
```
